# NN zero-D parameter comparison (VMR_aortas demo)

This notebook trains **junction and vessel neural networks** on four VMR aortas and evaluates on a held-out fifth geometry.
It then plots grouped bar charts of Poiseuille resistance **R**, stenosis coefficient **S**, and inductance **L**, comparing:

- **Standard** geometric 0D model (`geometric`)
- **Calibrated** ground truth (`BloodVesselJunction`)
- **Learned junctions** (`BloodVesselJunction_NN`)
- **Learned junctions + vessels** (`BloodVesselJunction_NN_plus_Vessel_NN`)
- **Learned vessels only** (`NN_vessel`)

**Requirements:** `pip install -e ".[dev,notebook]"`, JAX (`jax[cpu]`). **No svZeroDPlus** — calibrated JSONs are bundled under `data/zeroD/VMR_aortas/gen_loss/`.

Expected runtime on CPU: ~15–30 minutes (mostly NN training: 6 models × 3 R/S/L heads).

## Setup

```python
from __future__ import annotations

import json
import subprocess
import sys
from pathlib import Path

import numpy as np

from IPython.display import Image, display

from learn_lpns import get_pipeline_config, repo_root
from learn_lpns.data_processing.generate_split_indices import generate_geometry_split
from learn_lpns.tools.basic import load_dict
from learn_lpns.visualizations.plot_zero_d_parameter_bars import plot_zero_d_parameter_bars
from learn_lpns.zerod_calibration.modality_paths import (
    NN_JUNCTION_AND_VESSEL_SUFFIX,
    NN_JUNCTION_ONLY_SUFFIX,
    NN_VESSEL_ONLY_SUFFIX,
)
from learn_lpns.zerod_calibration.nn_inference import run_junction_inference, run_vessel_inference

SET_NAME = "VMR_aortas"
RUN_CONFIG = "gen_loss"
GEOMETRY_VARIANT = "bifurcations_EL"
JUNCTION_TYPE = "BloodVesselJunction"
NUM_GEOS = 5
VAL_GEO = "0076_1001"

GEOMETRIES = sorted(["0075_1001", "0076_1001", "0094_0001", "0095_0001", "0105_0001"])

cfg = get_pipeline_config(set_name=SET_NAME)
percent_train = cfg.split.data_processing_percent_train
seed = cfg.split.seed

train_geos, val_geos = generate_geometry_split(percent_train, seed, GEOMETRIES)
print(f"percent_train={percent_train}, seed={seed}")
print(f"Train ({len(train_geos)}): {train_geos}")
print(f"Val   ({len(val_geos)}): {val_geos}")
assert val_geos == [VAL_GEO], f"Expected val geo {VAL_GEO}, got {val_geos}"

ROOT = repo_root()
DATA_ROOT = ROOT / "data"
ZERO_D_BASE = DATA_ROOT / "zeroD" / SET_NAME / RUN_CONFIG
MODEL_DIR = (
    ROOT / "results" / "models" / "notebook_demo" / SET_NAME / f"{GEOMETRY_VARIANT}_trial_0"
)
VESSEL_MODEL_DIR = (
    ROOT / "results" / "models" / "notebook_demo" / SET_NAME / f"{GEOMETRY_VARIANT}_vessel_trial_0"
)
PLOT_DIR = ROOT / "results" / "notebook_demo" / "param_comparison" / RUN_CONFIG / SET_NAME / VAL_GEO
```

## 1 — Verify bundled calibrated data

```python
REQUIRED = (
    "bifurcations_EL_geometric_input.json",
    "bifurcations_EL_calibrated_output_BloodVesselJunction.json",
)

missing = []
for geo in GEOMETRIES:
    geo_dir = ZERO_D_BASE / geo
    for name in REQUIRED:
        path = geo_dir / name
        if not path.is_file():
            missing.append(str(path))

if missing:
    raise FileNotFoundError(
        "Missing bundled demo JSONs:\n  "
        + "\n  ".join(missing)
        + "\n\nSee data/README.md — run batch-zerod once locally to export, or pull latest main."
    )

print(f"OK: {len(GEOMETRIES)} geometries × {len(REQUIRED)} JSONs under {ZERO_D_BASE}")
```

## 2 — Build training artifacts (`ml_inputs`, `jax_arrays`, split)

```python
cmd = [
    sys.executable,
    "-m",
    "learn_lpns.data_processing.run_data_processing",
    "--set_name",
    SET_NAME,
    "--geometry_variant",
    GEOMETRY_VARIANT,
    "--run_config",
    RUN_CONFIG,
    "--percent_train",
    str(percent_train),
    "--seed",
    str(seed),
]
print(" ".join(cmd))
subprocess.run(cmd, cwd=ROOT, check=True)

jax_path = (
    DATA_ROOT
    / "jax_arrays"
    / SET_NAME
    / RUN_CONFIG
    / GEOMETRY_VARIANT
    / "all"
    / f"jax_arrays_num_geos_{NUM_GEOS}.pkl"
)
split_path = (
    DATA_ROOT
    / "split_indices"
    / SET_NAME
    / RUN_CONFIG
    / GEOMETRY_VARIANT
    / "all"
    / f"train_val_ind_{SET_NAME}_num_geos_{NUM_GEOS}"
)
assert jax_path.is_file(), jax_path
assert split_path.is_file(), split_path
print(f"jax_arrays: {jax_path}")
print(f"split:      {split_path}")
```

## 3 — Train junction and vessel NNs (3× R / S / L networks each)

```python
MODEL_DIR.mkdir(parents=True, exist_ok=True)
VESSEL_MODEL_DIR.mkdir(parents=True, exist_ok=True)

junction_cmd = [
    "learn-lpns-train",
    SET_NAME,
    str(NUM_GEOS),
    GEOMETRY_VARIANT,
    "--run_config",
    RUN_CONFIG,
    "--model_dir",
    str(MODEL_DIR),
]
print(" ".join(junction_cmd))
subprocess.run(junction_cmd, cwd=ROOT, check=True)

for i in range(3):
    model_path = MODEL_DIR / f"rri_{SET_NAME}_pred_{i}_model"
    assert model_path.is_file(), model_path
print(f"Junction models saved under {MODEL_DIR}")

vessel_cmd = [
    "learn-lpns-train",
    SET_NAME,
    str(NUM_GEOS),
    GEOMETRY_VARIANT,
    "--run_config",
    RUN_CONFIG,
    "--vessel",
    "--model_dir",
    str(VESSEL_MODEL_DIR),
]
print(" ".join(vessel_cmd))
subprocess.run(vessel_cmd, cwd=ROOT, check=True)

for i in range(3):
    model_path = VESSEL_MODEL_DIR / f"rri_{SET_NAME}_vessel_pred_{i}_model"
    assert model_path.is_file(), model_path
print(f"Vessel models saved under {VESSEL_MODEL_DIR}")
```

## 4 — Junction and vessel NN inference on held-out geometry

```python
val_dir = ZERO_D_BASE / VAL_GEO
geom_json = val_dir / "bifurcations_EL_geometric_input.json"
nn_json = val_dir / f"{GEOMETRY_VARIANT}_NN_{NN_JUNCTION_ONLY_SUFFIX}.json"
nn_junction_and_vessel_json = val_dir / f"{GEOMETRY_VARIANT}_NN_{NN_JUNCTION_AND_VESSEL_SUFFIX}.json"
nn_vessel_only_json = val_dir / f"{GEOMETRY_VARIANT}_NN_{NN_VESSEL_ONLY_SUFFIX}.json"

with open(geom_json) as f:
    nn_config = json.load(f)

jax_data = load_dict(str(jax_path))
run_junction_inference(
    jax_data_dict=jax_data,
    nn_config=nn_config,
    set_name=SET_NAME,
    geo_name=VAL_GEO,
    model_dir=str(MODEL_DIR),
    junction_type=JUNCTION_TYPE,
    quadratic_resistor=False,
)

with open(nn_json, "w") as f:
    json.dump(nn_config, f, indent=4)
print(f"Wrote {nn_json}")

with open(nn_json) as f:
    junction_nn_config = json.load(f)
junction_and_vessel_config, vessel_only_config = run_vessel_inference(
    junction_nn_config=junction_nn_config,
    variant_geometric_input=str(geom_json),
    set_name=SET_NAME,
    geometry_variant=GEOMETRY_VARIANT,
    model_dir=str(MODEL_DIR),
    quadratic_resistor=False,
    verbose=True,
)

with open(nn_junction_and_vessel_json, "w") as f:
    json.dump(junction_and_vessel_config, f, indent=4)
with open(nn_vessel_only_json, "w") as f:
    json.dump(vessel_only_config, f, indent=4)
print(f"Wrote {nn_junction_and_vessel_json}")
print(f"Wrote {nn_vessel_only_json}")
```

## 5 — Zero-D parameter comparison plot

```python
modality_json_paths = {
    "geometric": str(geom_json),
    JUNCTION_TYPE: str(val_dir / "bifurcations_EL_calibrated_output_BloodVesselJunction.json"),
    "BloodVesselJunction_NN": str(nn_json),
    "BloodVesselJunction_NN_plus_Vessel_NN": str(nn_junction_and_vessel_json),
    "NN_vessel": str(nn_vessel_only_json),
}

PLOT_DIR.mkdir(parents=True, exist_ok=True)
plot_path = plot_zero_d_parameter_bars(
    modality_json_paths,
    output_dir=str(PLOT_DIR),
    output_name=f"{GEOMETRY_VARIANT}_zero_d_parameter_bars.png",
    verbose=True,
)

if plot_path:
    display(Image(filename=plot_path))
else:
    print("Plot generation failed — check modality JSON paths.")
```

## 6 — Quantitative summary (MAE vs calibrated)

```python
PARAMS = ("R_poiseuille", "stenosis_coefficient", "L")


def _is_non_el_connector(name: str | None) -> bool:
    if not name:
        return True
    lower = name.lower()
    return "connector" in lower and "connectorel" not in lower


def load_modality_data(paths: dict[str, str]) -> dict[str, dict]:
    data = {}
    for mod, path in paths.items():
        with open(path) as f:
            data[mod] = json.load(f)
    return data


def param_vectors(mod_json: dict, labels: list[str], junction_outlet_map: dict) -> dict[str, np.ndarray]:
    vessels = {v.get("vessel_name"): v.get("zero_d_element_values", {}) for v in mod_json.get("vessels", [])}
    jmap = {j.get("junction_name"): j.get("junction_values", {}) for j in mod_json.get("junctions", [])}
    out = {p: [] for p in PARAMS}
    for name in labels:
        if name in vessels:
            z = vessels[name]
            for p in PARAMS:
                out[p].append(float(z.get(p)) if z.get(p) is not None else np.nan)
        else:
            jname, oi = junction_outlet_map[name]
            for p in PARAMS:
                vals = jmap.get(jname, {}).get(p)
                if isinstance(vals, list) and oi < len(vals):
                    out[p].append(float(vals[oi]))
                else:
                    out[p].append(np.nan)
    return {p: np.asarray(v, dtype=float) for p, v in out.items()}


def build_labels(mod_json: dict) -> tuple[list[str], dict]:
    vessels = mod_json.get("vessels", [])
    vessel_names = [
        v.get("vessel_name")
        for v in vessels
        if v.get("vessel_name") and not _is_non_el_connector(v.get("vessel_name"))
    ][:10]
    junction_outlet_labels = []
    junction_outlet_map = {}
    for j in mod_json.get("junctions", []):
        jname = j.get("junction_name")
        if not jname:
            continue
        for oi, vidx in enumerate(j.get("outlet_vessels", [])):
            vidx_int = int(vidx)
            out_name = vessels[vidx_int].get("vessel_name") if 0 <= vidx_int < len(vessels) else None
            if _is_non_el_connector(out_name):
                continue
            label = f"{jname}:out{oi}"
            junction_outlet_labels.append(label)
            junction_outlet_map[label] = (jname, oi)
    return vessel_names + junction_outlet_labels, junction_outlet_map


COMPARE_MODALITIES = (
    ("geometric", "geometric"),
    ("BloodVesselJunction_NN", "junction NN"),
    ("BloodVesselJunction_NN_plus_Vessel_NN", "junction+vessel NN"),
    ("NN_vessel", "vessel NN"),
)

modality_data = load_modality_data(modality_json_paths)
labels, jmap = build_labels(modality_data["geometric"])
calibrated = param_vectors(modality_data[JUNCTION_TYPE], labels, jmap)
vectors = {
    key: param_vectors(modality_data[key], labels, jmap) for key, _ in COMPARE_MODALITIES
}

header = f"{'Parameter':<22}" + "".join(f"{label:>18}" for _, label in COMPARE_MODALITIES[1:])
print(header)
print("-" * len(header))
for p in PARAMS:
    ref = calibrated[p]
    row = f"{p:<22}"
    for key, _ in COMPARE_MODALITIES[1:]:
        mae = float(np.nanmean(np.abs(vectors[key][p] - ref)))
        row += f"{mae:18.4g}"
    print(row)

mae_geo = {p: float(np.nanmean(np.abs(vectors["geometric"][p] - calibrated[p]))) for p in PARAMS}
mae_jv = {
    p: float(np.nanmean(np.abs(vectors["BloodVesselJunction_NN_plus_Vessel_NN"][p] - calibrated[p])))
    for p in PARAMS
}
print("\nJunction+vessel NN closer than geometric?")
for p in PARAMS:
    print(f"  {p}: {'yes' if mae_jv[p] < mae_geo[p] else 'no'}")

print(
    "\nInterpretation: lower MAE vs calibrated is better. "
    "Junction+vessel NN should beat geometric on R; exact rankings may vary with training seed."
)
```
